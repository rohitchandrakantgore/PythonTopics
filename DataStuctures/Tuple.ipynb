# Tuple

-   Same as List except it is immutable i.e. cannot perform any changes in the object
-   Read Only Version of List
-   Insertion Order Preserved
-   Duplicates Allowed
-   Heterogenous Objects Allowed
-   represent within ***parenethesis ()*** with comma seperator.
-   parenthesis optional

```python
# Parentheses are optional when creating a tuple

t1 = (10,20,30,40)
t2 = 10,20,30,40

print(t1,t2)
print(type(t1),type(t2))

# A single-value tuple requires a trailing comma
t3 = 10
t4 = 10,

print(type(t3), type(t4))
```

```python
# Tuple creation examples
t1 = ()
t2 = (10,20,30)
t3 = 10, 20, 30
lis = [10,20,30]
t4 = tuple(lis)

print(type(t1),type(t2),type(t3),type(t4))
```

## Accesing Elements of Tuple

```python
tup = (10,20,30,40)

# Access tuple elements by index
print(tup[0])
print(tup[-1])

# Access a range of elements with slicing
print(tup[1:3])
```

## Mathematical Operators for Tuple

```python
t1 = (10,20)
t2 = (30,40)

# Concatenate tuples with +
print(t1+t2)

# Repeat tuple contents with *
print(t1*3)
```

## Important Functions Of Tuple

```python
t = (10,20,30,40,10,20)

# Length of the tuple
print(len(t))

# Count occurrences of a value
print(t.count(10))

# Index of the first occurrence of a value
print(t.index(10))

# Sorted returns a list of tuple items in ascending order
print(sorted(t))

# Minimum and maximum values in the tuple
print(min(t))
print(max(t))
```

### Tuple Packing And Unpacking

```python
# Packing values into a tuple
a,b,c,d = 10,20,30,40
t = a,b,c,d
print(t)

# Unpacking tuple values into variables
a,b,c,d = t
print(a,b,c,d)
```

## Problems on Tuples

### Joining Tuples

```python
test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]

res = []
for sub in test_list:
    print(sub)
    if res and res[-1][0] == sub[0]:
        res[-1].extend(sub[1:])
    else:
        res.append([ele for ele in sub])
res = list(map(tuple, res))
print(res)
```

```python
test_list = [(5, 6), (5, 7), (5, 8), (6, 10), (7, 13)]
for i in range(len(test_list)-1):
    new = test_list[i] + test_list[i+1]
    print(new)
```

```python
dji1 = ('AAPL.US', 'IBM.US', 'MSFT.US')
dji2 = ('HD.US', 'GS.US', 'NKE.US')
result = dji1 + dji2
print(result)
```

```python
dji1 = ('AAPL.US', 'IBM.US', 'MSFT.US')
dji2 = ('HD.US', 'GS.US', 'NKE.US')
result = (dji1, dji2)
print(result)
```

```python
members = (('Kate', 23), ('Tom', 19))
members = (members[0], ('John', 26), members[1])
print(members)
```

```python
default = ('YES', 'NO', 'NO', 'YES', 'NO')
print(f"Number of occurrences: {default.count('YES')}")
```

```python
names = ('Monica', 'Tom', 'John', 'Michael')

# Sort tuple values alphabetically and convert back to a tuple
sorted_names = tuple(sorted(names))
print(sorted_names)
```

```python
info = (('Monica', 19), ('Tom', 21), ('John', 18))

# Sort by the second item in each tuple (age)
asc = tuple(sorted(info, key=lambda item: item[1]))
desc = tuple(sorted(info, key=lambda item: item[1], reverse=True))
print(f'Ascending: {asc}')
print(f'Descending: {desc}')
```

```python
stocks = (('Apple Inc', ('AAPL.US', 310)), ('Microsoft Corp', ('MSFT.US', 184)))

# Access a nested tuple element by index
print(stocks[0][1][0])
```
