# Control Flow in Python

# 1.  Conditional Statements
-   if
-   if-elif
-   if-else
-   if-elif-else


```python
# conditional statements control flow
# if conditional
number = 10
if number%2==0:
    print(number)

#if-elif conditional
number = 15
if number < 10:
    print("Number is less than 10")
elif number < 20:
    print("Number is less than 20")

# if-else conditional
number = 25
if number < 20:
    print("Number is less than 20")
else:
    print("Number is greater than or equal to 20")

# if-elif-else conditional
number = 30
if number < 20:
    print("Number is less than 20")
elif number < 40:
    print("Number is less than 40")
else:
    print("Number is greater than or equal to 40")
```

# 2. Transfer Statement
-   break
-   continue
-   pass

```python
# Transfer of control using break and continue statements
# break statement
for i in range(10):
    if i == 5:
        break
print("Loop exited at i =", i)

# continue statement
for i in range(10):
    if i % 2 == 0:
        continue
    print(i)
print("Loop completed")

# pass statement
for i in range(5):
    if i == 3:
        pass  # Do nothing for i == 3
    else:
        print(i)
print("Loop completed")
```

# 3. Iterative Statements
-   for
-   while

```python
# Iterative control using for and while loops
# for loop
for i in range(5):
    print(i)
print("Loop completed")

# while loop
count = 0
while count < 5:
    print(count)
    count += 1
print("Loop completed")
```

## Key points to remember

- Indentation defines the body of a conditional or loop.
- `range(start, stop, step)` excludes `stop`; a `while` loop needs a condition that eventually becomes false.
- `break` exits the nearest loop, `continue` skips to its next iteration, and `pass` is a no-op placeholder.
- A loop's optional `else` block runs only when the loop finishes normally, not when it exits with `break`.
- Conditions use truthiness: empty collections, `0`, `0.0`, `None`, and `False` are false; most other values are true.
