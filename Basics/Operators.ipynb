# Operators in Python

```python
# Arithmetic Operators
a = 10
b = 5
c = 2.5

d = a + b # addition
e = a - b # subtraction
f = a * b # multiplication
g = a / b # division
h = a // b # floor division
i = a % b # modulus
j = a ** c # exponentiation

print("Addition:", d)
print("Subtraction:", e)
print("Multiplication:", f)
print("Division:", g)
print("Floor Division:", h)
print("Modulus:", i)
print("Exponentiation:", j) 
```

```python
# Relational Operators
x = 10
y = 5
z = 10

print("x == y:", x == y) #checks if x is equal to y
print("x != y:", x != y) #checks if x is not equal to y
print("x > y:", x > y) #checks if x is greater than y
print("x < y:", x < y) #checks if x is less than y
print("x >= z:", x >= z) #checks if x is greater than or equal to z
print("x <= z:", x <= z) #checks if x is less than or equal to z
```

```python
# Equality Operators
m = 10
n = 10
o = 5

print("m == n:", m == n) #checks if m is equal to n
print("m == o:", m == o) #checks if m is equal to o
print("m != o:", m != o) #checks if m is not equal to o
```

```python
# Logical Operators
p = True
q = False

print("p and q:", p and q) #checks if both p and q are True
print("p or q:", p or q) #checks if either p or q is True
print("not p:", not p) #checks if p is False
```

```python
# Bitwise Operators
x = 10  # binary: 1010
y = 4   # binary: 0100
z = x & y  # bitwise AND
print("Bitwise AND:", z)

z = x | y  # bitwise OR
print("Bitwise OR:", z)

z = x ^ y  # bitwise XOR
print("Bitwise XOR:", z)

z = ~x     # bitwise NOT
print("Bitwise NOT:", z)

x = 10  # binary: 1010
z = x << 2  # left shift
print("Left Shift:", z)

z = x >> 2  # right shift
print("Right Shift:", z)
```

```python
# Assignment Operators
a = 10
a += 5  # equivalent to a = a + 5
print("After += 5:", a)

a -= 3  # equivalent to a = a - 3
print("After -= 3:", a)

a *= 2  # equivalent to a = a * 2
print("After *= 2:", a)

a /= 2  # equivalent to a = a / 2
print("After /= 2:", a)
```

```python
# Ternary Operator
age = 18
status = "Adult" if age >= 18 else "Minor"
print("Status:", status)
```

```python
# Identity Operators
x = [1, 2, 3]
y = x  # y references the same list as x
z = [1, 2, 3]  # z is a different list with the same contents

print("x is y:", x is y)  # True
print("x is z:", x is z)  # False
print("x == z:", x == z)  # True
```

```python
# Membership Operators
list1 = [1, 2, 3, 4, 5]
print("3 in list1:", 3 in list1)  # True
print("6 not in list1:", 6 not in list1)  # True
```

## Operator notes

- `==` compares values; `is` checks whether two names refer to the same object. Use `is None` for a `None` check, not for ordinary value equality.
- `and` and `or` short-circuit and return one of their operands; they do not always return a `bool`.
- Comparisons can be chained, as in `0 < value < 10`.
- `**` is exponentiation; `^` is bitwise XOR. `not` has different precedence from comparisons, so parentheses help make compound conditions clear.
- Augmented assignments such as `+=`, `-=`, `*=`, `/=`, `//=`, `%=`, and `**=` update a variable using an operation.
