# Keywords in Python

Keywords are reserved words with special meaning in Python syntax; they cannot be used as ordinary variable or function names. The exact set depends on the Python version.

Some contextual or *soft keywords* act as keywords only in specific syntax positions. Use the `keyword` module rather than memorizing a version-specific list.

```python
import keyword

print("Reserved keywords:", keyword.kwlist)
print("Soft keywords:", keyword.softkwlist)
print("Is 'if' a keyword?", keyword.iskeyword("if"))
print("Is 'match' a soft keyword?", keyword.issoftkeyword("match"))
```
