# Set

- A set is a mutable collection of unique, hashable objects.
- Sets are unordered collections: they do not support positional indexing or slicing, and their iteration order is not a contract.
- Use `{...}` for a non-empty set and `set()` for an empty set (`{}` creates an empty dictionary).
- Set operations include union, intersection, difference, and symmetric difference.
- `frozenset` is the immutable set type; it can itself be used as a set element or dictionary key.

### Creation Of Set

```python
# Typ1e1
set1 = {10,20,30,40}
print(type(set1), set1)

#type 2
list1 = [10,20,30]
t = 19,30,32
set2 = set(list1)
set3 = set(t)
print(set2, set3)

#type 3
set4= set(range(10))
print(set4)

# in case of empty set 
# type 4

set5 ={} # it is dict
set6 = set() # this is correct for empty set
print(type(set5),type(set6))
```

### Functions of Set

```python
# add() accepts exactly one element.
set1 = {10, 20, 30}
set1.add(500)
print(set1)

try:
    set1.add(11, 12)
except TypeError as error:
    print("add() accepts one element:", error)
```

```python
# update() adds elements from an iterable.
set1 = {10, 20, 30}
list1 = [11, 22, 33]
set1.update(list1, range(100, 110))
print(set1)

try:
    set1.update(10)
except TypeError as error:
    print("update() needs an iterable:", error)
```

```python
# copy - returns copy of the set , it is a cloned object.
set1 = {10,20,30}
set2 = set1.copy()
print(set2)
```

```python
# pop() removes and returns an arbitrary element; the set may be empty afterward.
set1 = {10, 20, 30, 40, 50, 60}
removed = set1.pop()
print("Removed:", removed)
print("Remaining:", set1)
```

```python
# remove() raises KeyError when the element is absent.
set1 = {10, 20, 30, 40, 50}
set1.remove(30)
print(set1)

try:
    set1.remove(30)
except KeyError as error:
    print("Element is not in the set:", error)
```

```python
# discard - removes specified element form the set, if element not present then it wont return any error
set1 = {10,20,30,40,50}
set1.discard(30)
print(set1)
set1.discard(30)
print(set1)
```

```python
# clear - removes all elements from set
set1 = {10,20,30,40,50}
set1.clear()
print(set1)
```

### Mathmatical Operations on Set

```python
# union - x.union(y) or x|y  - returns all elements present in both sets
set1 ={10,20,30,40,12,43}
set2= {10,20,30,50,60}
print(set1.union(set2))
print(set1|set2)
```

```python
# intersection - x.itersection(y) or x&y - return common elements present in both
set1 ={10,20,30,40,12,43}
set2= {10,20,30,50,60}
print(set1.intersection(set2))
print(set1&set2)
```

```python
# difference - x.difference(y) or x-y - returns elements present in x but not in y

set1 ={10,20,30,40,12,43}
set2= {10,20,30,50,60}
print(set1.difference(set2))
print(set1-set2)
```

```python
# symmetric difference - x.symmetriv_difference(y) or x^y - returns elements present in either x or y but not in both

set1 ={10,20,30,40,12,43}
set2= {10,20,30,50,60}
print(set1.symmetric_difference(set2))
print(set1^set2)
```

### Membership Operators

```python
set1 ={10,20,30,40,12,43}
print(10 not in set1)
print(10 in set1)
```

### Set relationships

- `<=` and `<` test subset and proper-subset relationships; `>=` and `>` test supersets.
- `isdisjoint(other)` checks whether two sets share no elements.
- Set membership is generally much faster than list membership on large collections, though exact timings depend on data and runtime.

```python
required = {"name", "email"}
provided = {"name", "email", "phone"}
print("Required fields are present:", required <= provided)
print("Has no overlapping fields:", required.isdisjoint({"phone", "address"}))
```
