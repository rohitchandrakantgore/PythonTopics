# String Data Type

A Python `str` is an immutable sequence of Unicode characters. String literals can use single quotes, double quotes, or triple quotes; use triple quotes when the text spans lines. Indexing and slicing return strings, and string methods return new strings rather than changing the original.

```python
# single quote and double quote are used to represent string literals in Python.
double_quote_string= "Python is a high-level, interpreted, general-purpose programming language." # double quotes
single_quote_string= 'Python is a high-level, interpreted, general-purpose programming language.' # single quotes

print(double_quote_string)
print(single_quote_string)
```

```python
# triple quotes are used to represent multi-line string literals in Python.
triple_quote_string = """Python is a high-level, interpreted, general-purpose programming language. 
Being a general-purpose language,it can be used to build almost any type of application with the right tools/libraries. Additionally, python supports objects, modules, threads, exception-handling,
and automatic memory management which help in modelling real-world problems and building applications to solve these problems."""
print(triple_quote_string)
print(len(triple_quote_string)) # to get all the no of characters or length of string
```

## Accessing the characters of string
-   By Using Index (+ve, -ve indexing)
-   By Using Slice Operator

```python
print(triple_quote_string[0])
print(triple_quote_string[10])
print(triple_quote_string[-20])

try:
    print(triple_quote_string[-10000])
except IndexError as error:
    print("Index outside the string:", error)
```

```python
# Slice syntax: [start:stop:step]; stop is excluded.
print(triple_quote_string[::])  # Entire string using default slice bounds
print(triple_quote_string[4:16])
print(triple_quote_string[-16:-4])
print(triple_quote_string[4:16:3])
print(triple_quote_string[::-1])  # Reverse
```

## Removing whitespace from a string

- `rstrip()` removes leading/trailing whitespace on the right.
- `lstrip()` removes whitespace on the left.
- `strip()` removes whitespace from both ends.
- These methods return new strings; they do not modify the original string.

```python
strip_example= " Python "
print(strip_example.rstrip()+"check")
print(strip_example.lstrip()+"check")
print(strip_example.strip()+"check")
```

## Find Substring

```python
ex = "Will Smith"

print("find missing:", ex.find("R"))  # -1 when not found
print("index found:", ex.index("W"))
try:
    ex.index("R")
except ValueError as error:
    print("index() missing substring:", error)

print("rfind:", ex.rfind("l"))
print("rfind missing:", ex.rfind("R"))  # -1 when not found
try:
    ex.rindex("R")
except ValueError as error:
    print("rindex() missing substring:", error)
```

## check type of characters in String

```python
ex = "Hi, hello"
print(ex.isalnum()) # A string is alpha-numeric if all characters in the string are alpha-numeric and there is at least one character in the string.
print(ex.isalpha()) # A string is alphabetic if all characters in the string are alphabetic and there is at least one character in the string.
print(ex.isascii()) # ASCII characters have code points in the range U+0000-U+007F. Empty string is ASCII too.
print(ex.isdecimal()) # A string is a decimal string if all characters in the string are decimal and there is at least one character in the string
print(ex.isidentifier()) # Call keyword.iskeyword(s) to test whether string s is a reserved identifier, such as "def" or "class".
print(ex.islower()) # A string is lowercase if all cased characters in the string are lowercase and there is at least one cased character in the string.
print(ex.isprintable()) # A string is printable if all of its characters are considered printable in repr() or if it is empty.
print(ex.isspace()) # A string is whitespace if all characters in the string are whitespace and there is at least one character in the string.
print(ex.istitle()) # In a title-cased string, upper- and title-case characters may only follow uncased characters and lowercase characters only cased ones.
print(ex.isupper()) # A string is uppercase if all cased characters in the string are uppercase and there is at least one cased character in the string
print(ex.isnumeric()) # A string is numeric if all characters in the string are numeric and there is at least one character in the string.
print(ex.isdigit()) # A string is a digit string if all characters in the string are digits and there is at least one character in the string.
```

#### WAP for
        input - a4b3c2
        output - aaaabbbcc

```python
import re

compressed = "a4b3c2"
if not re.fullmatch(r"(?:[A-Za-z]\d+)+", compressed):
    raise ValueError("Expected letter/count pairs, such as a4b3c2")

pairs = re.findall(r"([A-Za-z])(\d+)", compressed)
expanded = "".join(character * int(count) for character, count in pairs)
print(expanded)
```

```python
input_ = "AAASDSDSSSDDDFFFGG"
output=[]
for i in input_:
    if i not in output:
        output.append(i)
print("".join(output))
```

####
    Input = "AABBBCCCCCDDEFF"
    Output = {"A":2,"B":3,"C":5,"E":1,"F":2}

```python
input_ = "AABBBCCCCCDDEFF"
output={}

for i in input_:
    if i in output.keys():
        output[i]=output[i]+1
    else:
        output[i]=1
print(output)
```

## Formatting and encoding

- F-strings interpolate expressions: `f"Hello, {name}"`.
- A Python `str` is text; `bytes` is binary data. Encode text to bytes for storage or transmission, and decode bytes when the text encoding is known.
- `split()` returns a list of strings; `separator.join(parts)` combines strings and requires each part to be a string.

```python
name = "Rohit"
score = 95
print(f"{name} scored {score} points")

text = "cafe"
encoded_text = text.encode("utf-8")
print("Bytes:", encoded_text)
print("Decoded text:", encoded_text.decode("utf-8"))
```
