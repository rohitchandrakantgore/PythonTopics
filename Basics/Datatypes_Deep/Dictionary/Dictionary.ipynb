# Dictionary

- A dictionary maps unique keys to values; values may repeat and can have different types.
- Keys must be hashable (for example, strings, numbers, or tuples of hashable values). Lists and dictionaries cannot be keys.
- Dictionaries are mutable and dynamically sized. Insertion order is guaranteed in Python 3.7 and later.
- Access values by key, not by positional index or slice. Use `key in mapping` to test for a key; membership does not search values.
- `mapping.get(key, default)` is useful when a missing key should not raise `KeyError`.

### Create Dictionary

```python
#empty dictionary
dict1 = {}
dict2 = dict()
print(type(dict1), type(dict2))

dict3 = {1:"A",2:"B",3:"C"}
dict3[4]="D"
dict3[5] = "E"
print(type(dict3), dict3)
```

### Access Data from Dictionary

```python
dict1 = {1: 'A', 2: 'B', 3: 'C', 4: 'D', 5: 'E'}
print(dict1[2])

try:
    print(dict1[10])
except KeyError as error:
    print("Missing key:", error)

print("Default with get():", dict1.get(10, "not found"))
```

```python
dict1 = {1: 'A', 2: 'B', 3: 'C', 4: 'D', 5: 'E'}
for each in dict1:
    print(each, dict1[each])
```

### Update Dictionaries

```python
dict1 = {1: 'A', 2: 'B', 3: 'C', 4: 'D', 5: 'E'}
dict1[1]="Aa"
print(dict1)
```

### Delete Elements from Dictionary

```python
dict1 = {1: 'A', 2: 'B', 3: 'C', 4: 'D', 5: 'E'}
del dict1[1]
print(dict1)

try:
    del dict1[1]
except KeyError as error:
    print("Cannot delete a missing key:", error)
```

```python
# clear
dict1 = {1: 'A', 2: 'B', 3: 'C', 4: 'D', 5: 'E'}

# d.clear() - to remove all entries from dictionary
dict1.clear()
print(dict1)
```

```python
dict1 = {1: 'A', 2: 'B', 3: 'C', 4: 'D', 5: 'E'}
del dict1

try:
    print(dict1)
except NameError as error:
    print("The dictionary name was deleted:", error)
```

### FUnctions of dictionary

```python
# dict() - to create a dictionary
dict1 = dict()
dict2 = dict({1:"A"})
dict3 = dict([(1,"A"),(2,"B")])
print(dict1, dict2, dict3)
```

```python
# len() - return no of items in dictionary
dict3 = dict([(1,"A"),(2,"B")])
print(len(dict3))
```

```python
# clear() - to remove all elements
dict3 = dict([(1,"A"),(2,"B")])
dict3.clear()
print(dict3)
```

```python
# get()

## d.get(key) - if key is availbale return key else return None
dict3 = dict([(1,"A"),(2,"B")])
print(dict3.get(1))
print(dict3.get(10)) # None

## d.get(key, defaultvalue) - returns key is available else returns default value
dict3 = dict([(1,"A"),(2,"B")])
print(dict3.get(1,"One"))
print(dict3.get(10,"Ten")) # Ten
```

```python
# pop() removes the key and returns its value; a missing key raises KeyError.
dict3 = dict([(1, "A"), (2, "B")])
print(dict3.pop(1))

try:
    dict3.pop(10)
except KeyError as error:
    print("Missing key:", error)
```

```python
# popitem() removes and returns the most recently inserted (key, value) pair.
dict1 = {1: 'A', 2: 'B', 3: 'C', 4: 'D', 5: 'E'}
print(dict1.popitem())
```

```python
# keys() - returns all keys from dictionary
dict1 = {1: 'A', 2: 'B', 3: 'C', 4: 'D', 5: 'E'}
print(dict1.keys())
for key in dict1.keys():
    print(key)
```

```python
# values() - returns all values from dictionary
dict1 = {1: 'A', 2: 'B', 3: 'C', 4: 'D', 5: 'E'}
print(dict1.values())
for value in dict1.values():
    print(value)
```

```python
# items() returns a dynamic view of the dictionary's key-value pairs.
dict1 = {1: 'A', 2: 'B', 3: 'C', 4: 'D', 5: 'E'}
print(dict1.items())
for key, value in dict1.items():
    print(key, value)
```

```python
# copy() - to create exact copy of dictionary
dict1 = {1: 'A', 2: 'B', 3: 'C', 4: 'D', 5: 'E'}
dict_new = dict1.copy()
print(dict_new)
```

## More dictionary patterns

- `update(other)` inserts new key-value pairs and replaces values for existing keys.
- `setdefault(key, default)` returns the existing value, or inserts and returns the default when the key is missing.
- `key in mapping` checks keys; use `value in mapping.values()` when the intention is to search values.
- `copy()` makes a shallow copy: nested mutable values are still shared.

```python
settings = {"theme": "light", "font_size": 12}
settings.update({"theme": "dark", "language": "en"})
settings.setdefault("show_tips", True)
settings.setdefault("theme", "light")  # Existing value is kept.

print(settings)
```
