# Python Lists - Additional Concepts


## 1. Creating Lists from User Input

Python allows users to create lists dynamically by taking input during program execution.

---

## Using eval()

### Purpose

`eval()` evaluates the input string as a Python expression and converts it into the corresponding Python object.

### Syntax

```python
list_name = eval(input("Enter Elements: "))
```

### Characteristics

- Allows direct creation of lists from user input.
- Can convert string representation of lists into actual lists.
- Supports Python expressions.
- Returns the evaluated object.

### Important Points

- `eval()` should be avoided with untrusted user input because it can execute arbitrary Python code.
- Safer alternatives include:
  - `ast.literal_eval()`
  - Taking individual inputs and converting them manually.

---


````python
#eval()
list_elements = eval(input("enter list list_elements: "))
list_elements
````

# 2. Creating Lists Using String Operations

## split()

### Purpose

Converts a string into a list by separating elements based on a delimiter.

### Syntax

```python
string.split(separator)
```

### Characteristics

- Returns a new list.
- Original string remains unchanged.
- Default separator is whitespace.
- Commonly used for converting sentences into lists of words.

### Example Use Cases

- Processing text data
- Reading user input
- Tokenizing sentences

---

````python
Name = "A B C"
named_list = Name.split(' ')
print(named_list)  # Output: ['A', 'B', 'C']
````

# 3. Mathematical Operations on Lists

Python lists support some operations using operators.

---

# Addition Operator (+)

### Purpose

Combines two lists together.

### Characteristics

- Performs list concatenation.
- Creates a new list.
- Original lists remain unchanged.
- Maintains element order.

### Time Complexity

```
O(n+m)
```

where:

- n = size of first list
- m = size of second list

---

# Multiplication Operator (*)

### Purpose

Repeats list elements multiple times.

### Characteristics

- Creates a new list.
- Does not modify the original list.
- Repeats references of elements.
- Useful for initializing lists.

### Time Complexity

```
O(n*k)
```

where:

- n = list size
- k = repetition count

---

# Length Function (len())

### Purpose

Returns the number of elements present in a list.

### Syntax

```python
len(list)
```

### Characteristics

- Counts only top-level elements.
- Does not modify the list.
- Executes efficiently.

### Time Complexity

```
O(1)
```

---

# Minimum and Maximum Operations

## min()

Returns the smallest element from the list.

## max()

Returns the largest element from the list.

### Characteristics

- Elements must be comparable.
- Does not modify the list.
- Performs linear scanning.

### Time Complexity

```
O(n)
```

---

# Sum Operation

## sum()

### Purpose

Calculates the total of numeric elements.

### Characteristics

- Works mainly with numeric lists.
- Does not modify the list.
- Performs sequential addition.

### Time Complexity

```
O(n)
```

---

````python
numbers = [10, 20, 30, 40, 50]

#addition
numbers += [60, 70]

print("numbers after addition:", numbers)  # Output: [10, 20, 30, 40, 50, 60, 70]

#multiplication
numbers *= 2
print("numbers after multiplication:", numbers)  # Output: [10, 20, 30, 40, 50, 60, 70, 10, 20, 30, 40, 50, 60, 70]

#len() function
print("length of numbers:", len(numbers))  # Output: 14


#min() and max() functions
print("minimum value in numbers:", min(numbers))  # Output: 10
print("maximum value in numbers:", max(numbers))  # Output: 70

#sum() function
print("sum of numbers:", sum(numbers))  # Output: 560
````

# 4. Membership Operations in Lists

Membership operators are used to check whether an element exists inside a list.

Python provides two membership operators:

---

# in Operator

### Purpose

Checks whether a value exists in the list.

### Characteristics

- Returns True if element exists.
- Returns False if element does not exist.
- Searches sequentially.

### Time Complexity

```
O(n)
```

---

# not in Operator

### Purpose

Checks whether a value does not exist in the list.

### Characteristics

- Opposite of `in`.
- Returns True when element is absent.
- Returns False when element exists.

### Time Complexity

```
O(n)
```

---

````python
numbers = [10, 20, 30, 40, 50]

#in operator
print(20 in numbers)  # Output: True
print(100 in numbers)  # Output: False

#not in operator
print(20 not in numbers)  # Output: False
print(100 not in numbers)  # Output: True
````

# 5. List Comparison

Python allows comparison between lists using comparison operators.

Supported operators:

- ==
- !=
- <
- >
- <=
- >=

---

# Equality Comparison (==)

### Purpose

Checks whether two lists contain the same elements in the same order.

### Characteristics

- Compares elements one by one.
- Order matters.
- Nested elements are compared recursively.

---

# Relational Comparison

Lists can be compared using:

- less than (`<`)
- greater than (`>`)
- less than or equal (`<=`)
- greater than or equal (`>=`)

### Characteristics

- Uses lexicographical comparison.
- Compares elements from left to right.
- First different element decides the result.

---

````python
numbers = [10, 20, 30, 40, 50]

#equality comparison
print(numbers == [10, 20, 30, 40, 50])  # Output: True
print(numbers == [10, 20, 30])  #Output: False

#relational comparisons
print(numbers < [10, 20, 30, 40, 50, 60])  # Output: True
print(numbers > [10, 20, 30])  # Output: True
````

# 6. List Comprehension

## Introduction

List comprehension provides a concise way to create lists by applying expressions to elements of an iterable.

It is an alternative to traditional loops for creating lists.

---

## Syntax

```python
[expression for item in iterable]
```

---

## With Condition

```python
[expression for item in iterable if condition]
```

---

## Characteristics

- Creates a new list.
- More readable for simple transformations.
- Reduces the amount of code.
- Internally uses iteration.
- Supports conditions.
- Supports nested loops.

---

# Advantages of List Comprehension

- Cleaner syntax.
- Faster execution compared to some traditional loop approaches.
- Useful for transformations and filtering.
- Improves code readability when used appropriately.

---

# Disadvantages of List Comprehension

- Complex expressions reduce readability.
- Nested comprehensions can become difficult to maintain.
- Not suitable for operations with many side effects.

---


````python
#create list with list comprehension 
squares = [x**2 for x in range(1, 6)]
print("squares:", squares)  # Output: [1, 4, 9, 16, 25] 

#list comprehension with condition
even_squares = [x**2 for x in range(1, 11) if x % 2 == 0]
print("even squares:", even_squares)  # Output: [4, 16, 36, 64, 100]
````

# 8. List Aliasing

## Introduction

List aliasing occurs when multiple variables refer to the same list object.

---

## Characteristics

- Changes through one reference affect all references.
- Happens because lists are mutable objects.
- Different from copying a list.

---

# 9. Shallow Copy vs Deep Copy

## Shallow Copy

Creates a new list object but keeps references to nested objects.

---

## Deep Copy

Creates a completely independent copy including nested objects.

# 11. Common List Interview Concepts

## Mutable vs Immutable

Lists are mutable, meaning:

- Elements can be changed after creation.
- Size can increase or decrease.
- Methods modify the original object.

---

## List vs Tuple

| List | Tuple |
|------|-------|
| Mutable | Immutable |
| More memory usage | Less memory usage |
| Slower | Faster |
| More methods | Limited methods |
| Uses [] | Uses () |

---

## List vs Set

| List | Set |
|------|-----|
| Maintains order | No guaranteed order |
| Allows duplicates | Removes duplicates |
| Index based access | No indexing |
| Slower searching | Faster membership testing |
