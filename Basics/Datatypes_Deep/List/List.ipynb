# Python Lists: Core and Additional Concepts

A practical guide to creating, reading, changing, comparing, and copying Python lists. Each topic includes key points and a small example.

## Learning goals
- Recognize list behavior and common creation patterns.
- Access, traverse, and modify list elements.
- Use list methods, operators, built-ins, and comprehensions.
- Understand nested lists, aliasing, and copy depth.

## 1. List basics and creation

A list is an ordered, mutable sequence written with square brackets.

**Key points**
- Preserves insertion order and allows duplicate values.
- Can contain different data types, including other lists and `None`.
- Can grow or shrink; supports indexing, slicing, and iteration.
- `list(range(...))` creates a list of numbers; `split()` creates a list of text parts.
- Avoid `eval()` for input: it can execute arbitrary code. Use `ast.literal_eval()` for Python literals, or parse individual values.

```python
import ast

numbers = [10, 20, 30, 40]
names = ["Alice", "Bob", "Charlie"]
mixed = [10, "Python", 3.14, True]
empty = []
from_range = list(range(5))
words = "Python lists are useful".split()

# Safely parse a string containing a Python literal.
parsed_values = ast.literal_eval("[1, 2, 3]")

print(numbers, names, mixed, empty)
print(from_range, words, parsed_values)
```

## 2. Indexing and slicing

**Key points**
- Positive indexes start at `0`; negative indexes count backward from `-1`.
- Slice syntax is `items[start:stop:step]`; the stop index is excluded.
- Omitting start or stop uses the beginning or end of the list.
- A slice returns a new, shallow list; an invalid single index raises `IndexError`.

```python
items = [10, 20, 30, 40, 50]

print(items[0])       # first item
print(items[-1])      # last item
print(items[1:4])     # indexes 1, 2, and 3
print(items[::2])     # every second item
print(items[::-1])    # reversed copy
```

## 3. Traversing a list

**Key points**
- Iterate over values directly when the index is not needed.
- Use `enumerate()` when both index and value are useful.
- A `while` loop can traverse by index, but remember to advance the index.

```python
colors = ["red", "green", "blue"]

for color in colors:
    print(color)

for index, color in enumerate(colors):
    print(index, color)
```

```python
values = [1, 2, 3, 4]
index = 0

while index < len(values):
    print("Index:", index, "Value:", values[index])
    index += 1
```

## 4. Inspecting and modifying lists

**Useful methods**
- `append(value)`: add one value to the end.
- `insert(index, value)`: add a value at a position.
- `extend(iterable)`: add each value from an iterable.
- `remove(value)`: remove the first matching value; raises `ValueError` if absent.
- `pop([index])`: remove and return an item; defaults to the last item.
- `clear()`: remove all items.
- `count(value)` counts matches; `index(value)` finds the first match.

Most mutating list methods change the existing list in place and return `None`.

```python
scores = [80, 95, 80]
scores.append(88)
scores.insert(1, 91)
scores.extend([72, 84])
print("Updated:", scores)
print("Count of 80:", scores.count(80))
print("Index of 95:", scores.index(95))
scores.remove(72)
last_score = scores.pop()
print("Removed:", last_score)
print("Remaining:", scores)
```

```python
values = [10, 20, 10]
print("Count of 10:", values.count(10))

try:
    values.index(99)
except ValueError:
    print("99 is not in the list")

alias = values
values.clear()
print("Cleared list:", values)
print("Alias sees the same cleared list:", alias)
```

## 5. Reversing and sorting

**Key points**
- `reverse()` reverses a list in place; it does not sort it.
- `sort()` sorts a list in place and returns `None`.
- `sorted(items)` returns a new sorted list and leaves the input unchanged.
- Use `reverse=True` for descending order and `key=` for a custom sort rule.
- Values being compared during a sort need to be mutually comparable.

```python
values = [4, 1, 9, 2]
ascending = sorted(values)
descending = sorted(values, reverse=True)
values.sort()
print("Sorted copy:", ascending)
print("Descending copy:", descending)
print("Original after in-place sort:", values)
```

## 6. Operators and common built-ins

**Operators**
- `left + right` concatenates lists into a new list.
- `items * count` repeats the list into a new list. Repeated nested mutable items are references to the same objects.

**Built-ins**
- `len(items)` returns the number of top-level elements.
- `min()` and `max()` find extreme values; `sum()` totals numeric values.
- `any()` checks whether at least one item is truthy; `all()` checks whether every item is truthy.
- `zip()` pairs values from iterables; `enumerate()` pairs indexes and values.
- `len()` is $O(1)$; membership, `min()`, `max()`, and `sum()` are generally $O(n)$. List concatenation and repetition scale with the resulting number of elements.

```python
left = [10, 20, 30]
right = [40, 50]
combined = left + right
repeated = left * 2

print("Combined:", combined)
print("Repeated:", repeated)
print("Length / min / max / sum:", len(combined), min(combined), max(combined), sum(combined))
print("Any positive?", any(value > 0 for value in combined))
print("All positive?", all(value > 0 for value in combined))
print("Pairs:", list(zip(["A", "B"], [1, 2])))
```

## 7. Membership and comparison

**Key points**
- `value in items` checks presence; `value not in items` checks absence. A list membership check is generally $O(n)$.
- `==` checks whether contents and order match; `!=` checks whether they differ.
- Relational comparisons use lexicographic order: compare left to right until a differing pair decides the result.
- Compared elements must support the relevant comparison operation.

```python
letters = ["A", "B", "C"]
print("B" in letters)
print("Z" not in letters)
print(letters == ["A", "B", "C"])
print([1, 2, 9] < [1, 3])  # first differing values: 2 < 3
```

## 8. List comprehensions

A comprehension builds a new list by transforming and optionally filtering an iterable.

**Patterns**
- Transform: `[expression for item in iterable]`
- Filter: `[expression for item in iterable if condition]`
- Use comprehensions for clear, compact transformations; use a regular loop when logic is complex or has side effects.

```python
squares = [number ** 2 for number in range(1, 6)]
even_squares = [number ** 2 for number in range(1, 11) if number % 2 == 0]
print("Squares:", squares)
print("Even squares:", even_squares)
```

## 9. Nested lists

A nested list contains other lists and can represent rows, grids, or grouped data.

**Key points**
- Use chained indexes such as `grid[row][column]` to access nested values.
- `len(outer_list)` counts only its top-level elements.
- Nested mutable values matter when copying: a shallow copy still shares those inner objects.

```python
grid = [[1, 2, 3], [4, 5, 6]]
print("First row:", grid[0])
print("Second row, third value:", grid[1][2])
print("Number of rows:", len(grid))
```

## 10. Aliasing and copying

**Key points**
- Assignment such as `second = first` creates an alias: both names refer to the same list.
- `first.copy()` and `first[:]` make shallow copies: the outer lists are distinct, but nested objects are shared.
- `copy.deepcopy(first)` recursively copies nested objects when independent nested data is needed.
- Lists are mutable; changing a shared object is visible through every reference to it.

```python
import copy

original = [[1, 2], [3, 4]]
alias = original
shallow_copy = original.copy()
deep_copy = copy.deepcopy(original)

original[0][0] = 99
print("Original:", original)
print("Alias sees change:", alias)
print("Shallow copy shares inner list:", shallow_copy)
print("Deep copy remains independent:", deep_copy)
```

## 11. Choosing a collection

| Collection | Ordered | Mutable | Duplicates | Typical use |
|---|---|---|---|---|
| List | Yes | Yes | Yes | Changeable sequence with indexing |
| Tuple | Yes | No | Yes | Fixed sequence of values |
| Set | No positional indexing | Yes | No | Unique values and fast membership on average |

Choose based on the behavior needed, not on a blanket assumption that one type is always faster.

- Methods such as `append()`, `sort()`, and `reverse()` mutate the list.
- `+` concatenates; `*` repeats; `in` checks membership.
- A shallow copy does not recursively copy nested values.
