# Python List

## Introduction

A **list** is a built-in **sequence data type** in Python used to store an **ordered collection of elements**. It is one of the most versatile and frequently used data structures because it allows storing multiple values in a single variable.

Unlike many programming languages where arrays are of fixed size, Python lists are **dynamic**, meaning they can grow or shrink during program execution. Lists are also **mutable**, allowing elements to be added, removed, or modified after creation.

Lists can store elements of the **same or different data types**, making them suitable for representing a wide variety of real-world data such as student records, shopping carts, product inventories, and more.

---

## Characteristics of a List

- **Sequence Data Type** – Stores an ordered collection of elements.
- **Ordered** – Maintains the order in which elements are inserted.
- **Insertion Order Preserved** – Elements are retrieved in the same order they were added.
- **Mutable** – Elements can be added, removed, or modified after the list is created.
- **Dynamic in Size** – Automatically grows or shrinks as elements are added or removed.
- **Allows Duplicate Elements** – Multiple occurrences of the same value are permitted.
- **Supports Heterogeneous Data** – Can store elements of different data types in the same list.
- **Index-Based Access** – Elements can be accessed using positive and negative indexes.
- **Supports Slicing** – Allows extracting a portion of the list using slice notation.
- **Iterable** – Can be traversed using loops such as `for` and `while`.
- **Supports Nested Lists** – A list can contain one or more lists as its elements.
- **Can Store Any Python Object** – Supports primitive types, collections, functions, classes, and custom objects.
- **Resizable** – Memory is managed automatically; no need to specify the size in advance.

---

## Advantages

- Easy to create and use.
- Supports a wide variety of operations and built-in methods.
- Suitable for storing heterogeneous data.
- Dynamic memory allocation.
- Rich support for indexing, slicing, and iteration.
- Ideal for applications where data changes frequently.

---

## Limitations

- Searching for an element takes **O(n)** time in the worst case.
- Inserting or deleting elements in the middle requires shifting elements, making it **O(n)**.
- Consumes more memory than immutable data structures like tuples.
- Not suitable when fixed-size or immutable data is required.

# Creating Lists

A list can be created in multiple ways depending on the requirement. Python provides a simple and flexible syntax for creating lists containing zero or more elements.

## Syntax

```python
list_name = [element1, element2, element3, ...]
```

Elements are enclosed within **square brackets (`[]`)** and separated by commas.


## Creating an Empty List

An empty list contains no elements and can be created using empty square brackets.

````python
empty_list = []
print("Empty list:", empty_list)
````

Alternatively, the `list()` constructor can also be used.

````python
empty_list = list()
print("Empty list:", empty_list)
````

## Creating a List with Elements

A list can store one or more elements.

````python
numbers = [10, 20, 30, 40]
print("List of numbers:", numbers)
````

## Creating a List with Different Data Types

Lists support heterogeneous data, allowing different types of objects to be stored together.


````python
data = [101, "Python", 95.5, True]
data
````

## Creating a Nested List

A list can contain one or more lists as its elements.

````python
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
matrix
````

## Creating a List from an Iterable

The `list()` constructor converts any iterable object into a list.


````python
# From a String
letters = list("Python")
print("List from string:", letters)

### From a Tuple
numbers = list((10, 20, 30))
print("List from tuple:", numbers)

### From a Set
numbers = list({10, 20, 30})
print("List from set:", numbers)
##*Note:** Since sets are unordered, the resulting order may vary.


### From a Range
numbers = list(range(1, 6))
print("List from range:", numbers)
````

# Accessing Elements

Elements in a list are accessed using **indexing** or **slicing**.

Python supports both **positive indexing** and **negative indexing**, making it easy to retrieve elements from either end of the list.

---

## 1. Accessing Elements Using Index

Each element in a list has a unique position called an **index**.

### Positive Indexing

- Starts from **0**.
- Traverses from **left to right**.
- The first element has index `0`.

### Negative Indexing

- Starts from **-1**.
- Traverses from **right to left**.
- The last element has index `-1`.

### Invalid Index

Attempting to access an index outside the valid range raises an `IndexError`.


````python
fruits = ["Apple", "Banana", "Mango", "Orange"]

##positive indexing
print("Fruit 1:", fruits[0])
print("Fruit 2:", fruits[1])

##negative indexing
print("Last Fruit:", fruits[-1])
print("Second Last Fruit:", fruits[-2])

##invalid index
print("Invalid Index Fruit:", fruits[10])  # This will raise an IndexError
````

## 2. Accessing Elements Using Slicing

Slicing is used to retrieve a portion of a list.

### Syntax

```python
list[start : stop : step]
```

---

### Slice Parameters

| Parameter | Description | Default Value |
|-----------|-------------|---------------|
| `start` | Starting index (inclusive) | `0` |
| `stop` | Ending index (exclusive) | `len(list)` |
| `step` | Number of positions to move | `1` |

### Common Slicing Examples

````python
#### Retrieve a Range of Elements
numbers = [10, 20, 30, 40, 50]
print("Numbers from index 1 to 3:", numbers[1:4])  # Retrieves elements at index 1, 2, and 3

#### From Beginning to a Position
print("Numbers from start to index 2:", numbers[:3])  # Retrieves elements at index 0, 1, and 2

#### From a Position to the End
print("Numbers from index 2 to end:", numbers[2:])  # Retrieves elements at index 2, 3, and 4

#### Entire List
print("Entire list:", numbers[:])  # Retrieves all elements in the list

#### Every Second Element
print("Every second element:", numbers[::2])  # Retrieves every second element

#### Reverse the List
print("Reversed list:", numbers[::-1])  # Retrieves elements in reverse order

````

## Important Notes

- The **start index is inclusive**.
- The **stop index is exclusive**.
- Omitting `start` begins from index `0`.
- Omitting `stop` goes until the last element.
- Omitting `step` uses a default value of `1`.
- A negative `step` traverses the list in reverse order.

## Summary

| Method | Purpose |
|---------|---------|
| `list[index]` | Access a single element |
| `list[-index]` | Access elements from the end |
| `list[start:stop]` | Retrieve a range of elements |
| `list[start:stop:step]` | Retrieve elements with a custom step |
| `list[::-1]` | Reverse the list |

# Traversing Lists

Traversing a list means **visiting each element of the list one by one**. Python provides multiple ways to iterate over a list depending on the requirement.

The most common approaches are:

- Using a `for` loop
- Using a `while` loop
- Using `range()`
- Using `enumerate()`
- Traversing in reverse order

---


## 1. Using a `for` Loop

The `for` loop is the simplest and most Pythonic way to traverse a list. It directly iterates over each element without requiring an index.

### Syntax

```python
for element in list_name:
    # statements
```

````python
numbers = [10, 20, 30, 40]

for num in numbers:
    print(num)
````

## 2. Using a `while` Loop

A `while` loop is useful when you need explicit control over the index.

### Syntax

```python
i = 0

while i < len(list_name):
    # statements
    i += 1
```

````python
numbers = [10, 20, 30, 40]
i = 0

while i < len(numbers):
    print(numbers[i])
    i += 1
````

## 3. Using `range()`

Use `range()` when the index is required.


````python
numbers = [10, 20, 30]

for i in range(len(numbers)):
    print(i, numbers[i])
````

## 4. Using `enumerate()`

`enumerate()` returns both the index and the corresponding element. This is the recommended approach when both values are needed.

### Syntax

```python
enumerate(iterable, start=0)
```


````python
numbers = [10, 20, 30]

for index, value in enumerate(numbers):
    print(index, value)
````

## 5. Traversing in Reverse Order

Python provides two common ways to iterate in reverse.


````python
### Using `reversed()`

numbers = [10, 20, 30]
for num in reversed(numbers):
    print(num)
````

````python
### Using Slicing

numbers = [10, 20, 30]
for num in numbers[::-1]:
    print(num)
````

## Best Practices

- Use a **`for` loop** for normal traversal.
- Use **`enumerate()`** when both index and value are required.
- Use **`while`** only when manual control over the index is necessary.
- Prefer **`reversed()`** over slicing when simply iterating in reverse.

---

## Summary

| Method | Best Use Case |
|---------|---------------|
| `for` | Iterate over values |
| `while` | Manual index control |
| `range()` | Index-based iteration |
| `enumerate()` | Index and value together |
| `reversed()` | Reverse traversal |

# Python List Methods

## Introduction

Python lists are mutable sequence data structures that provide built-in methods to perform various operations such as:

- Adding elements
- Removing elements
- Searching elements
- Sorting data
- Reversing elements
- Copying lists
- Counting occurrences

List methods modify the existing list object unless explicitly mentioned otherwise.

Since lists are mutable, most methods perform changes **in-place** and return `None`.

---

# List Method Categories

Python list methods can be categorized into:

1. Adding Elements
2. Removing Elements
3. Searching and Counting
4. Ordering and Rearranging
5. Copying Lists
6. Utility Operations

---

# 1. Adding Elements

## append()

### Purpose

Adds a single element at the end of the list.

### Syntax

```python
list.append(element)
```

### Characteristics

- Adds exactly one element.
- Maintains the existing order.
- Modifies the original list.
- Increases list length by one.
- Accepts any data type including another list.

### Return Value

Returns `None`.

---

# extend()

### Purpose

Adds multiple elements from an iterable to the end of the list.

### Syntax

```python
list.extend(iterable)
```

### Characteristics

- Adds elements individually from another iterable.
- Commonly used to merge two lists.
- Accepts lists, tuples, strings, sets, and other iterables.
- Modifies the original list.
- Does not create a new list.

### Return Value

Returns `None`.

---

# insert()

### Purpose

Inserts an element at a specific index position.

### Syntax

```python
list.insert(index, element)
```

### Characteristics

- Maintains existing elements by shifting them.
- Allows insertion at any valid position.
- If index is greater than list length, inserts at the end.
- If index is negative, insertion happens relative to the end.

### Return Value

Returns `None`.

---

````python
numbers = [10,20,30]
print("original list:", numbers)

## append(element) - Adds an element to the end of the list
numbers.append(40)
print("After append:", numbers)

#extend(elements) - Adds multiple elements to the end of the list
numbers.extend([50, 60])
print("After extend:", numbers)

#insert(index, element) - Inserts an element at a specific index
numbers.insert(2, 25)  # Insert 25 at index 2
print("After insert:", numbers)
````

# 2. Removing Elements

# remove()

### Purpose

Removes the first occurrence of a specified value.

### Syntax

```python
list.remove(value)
```

### Characteristics

- Searches from the beginning.
- Removes only the first matching element.
- Raises an error if the element does not exist.
- Modifies the original list.

### Return Value

Returns `None`.

---

# pop()

### Purpose

Removes and returns an element from a specific position.

### Syntax

```python
list.pop(index)
```

### Characteristics

- Removes the element at the given index.
- Returns the removed element.
- If no index is provided, removes the last element.
- Commonly used with stack operations.

### Return Value

Returns the removed element.

---

# clear()

### Purpose

Removes all elements from the list.

### Syntax

```python
list.clear()
```

### Characteristics

- Deletes all list elements.
- Keeps the list object available.
- List length becomes zero.
- Does not delete the list variable.

### Return Value

Returns `None`.

---

````python
numbers = [10, 20, 30, 40, 50]
print("Original list:", numbers)

# remove(element) - Removes the first occurrence of an element from the list
numbers.remove(30)  # Removes the first occurrence of 30
print("After remove:", numbers)

# pop(index) - Removes and returns the element at a specific index
popped_element = numbers.pop(2)  # Removes and returns the element at index 2
print("After pop:", numbers)
print("Popped element:", popped_element)

# clear() - Removes all elements from the list
numbers.clear()
print("After clear:", numbers)
````

# 3. Searching and Counting

# index()

### Purpose

Finds the position of the first occurrence of a value.

### Syntax

```python
list.index(value)
```

### Characteristics

- Searches from left to right.
- Returns the index position.
- Can specify starting and ending positions.
- Raises an error if value is not found.

### Return Value

Returns integer index position.

---

# count()

### Purpose

Counts the number of occurrences of a value.

### Syntax

```python
list.count(value)
```

### Characteristics

- Searches the complete list.
- Does not modify the list.
- Useful for frequency calculations.

### Return Value

Returns integer count.

---

````python
numbers = [10, 20, 30, 40, 50]
print("Original list:", numbers)

#index(element) - Returns the index of the first occurrence of an element
index = numbers.index(30)  # Returns the index of the first occurrence of 30
print("Index of 30:", index)

#count(element) - Returns the number of occurrences of an element
count = numbers.count(20)  # Returns the count of 20 in the list
print("Count of 20:", count)

````

# 4. Ordering and Rearranging

# sort()

### Purpose

Sorts elements of the list in ascending or descending order.

### Syntax

```python
list.sort()
```

### Characteristics

- Performs sorting in-place.
- Modifies the original list.
- Supports custom sorting using key functions.
- Supports reverse ordering.
- Elements must be comparable.

### Return Value

Returns `None`.

---

# reverse()

### Purpose

Reverses the order of elements in the list.

### Syntax

```python
list.reverse()
```

### Characteristics

- Reverses elements in-place.
- Does not perform sorting.
- Maintains existing relationships between elements.

### Return Value

Returns `None`.

---

````python
numbers = [10, 20, 30, 40, 50]
print("Original list:", numbers)

#sort() - Sorts the list in ascending order
numbers.sort()
print("After sort:", numbers)

numbers.sort(reverse=True)  # Sorts the list in descending order
print("After sort in descending order:", numbers)

#reverse() - Reverses the order of elements in the list
numbers.reverse()
print("After reverse:", numbers)
````

# 5. Copying Lists

# copy()

### Purpose

Creates a shallow copy of the list.

### Syntax

```python
new_list = list.copy()
```

### Characteristics

- Creates a new list object.
- Copies references of elements.
- Changes in the copied list do not affect the original list for direct elements.
- Nested objects are still shared because it is a shallow copy.

### Return Value

Returns a new list object.

````python
numbers = [10, 20, 30, 40, 50]
print("Original list:", numbers)

#copy() - Returns a shallow copy of the list
numbers_copy = numbers.copy()
print("Copied list:", numbers_copy)
````

# Common Interview Points

## append() vs extend()

### append()

- Adds one object.
- Treats input as a single element.

### extend()

- Adds elements individually.
- Requires an iterable.

---

## remove() vs pop()

### remove()

- Removes using value.
- Does not return removed element.

### pop()

- Removes using index.
- Returns removed element.

---

## sort() vs sorted()

### sort()

- List method.
- Changes original list.
- Returns None.

### sorted()

- Built-in function.
- Creates a new sorted list.
- Original list remains unchanged.

---

## copy() vs Assignment

### Assignment

Creates another reference to the same list object.

### copy()

Creates a separate list object with copied references.


# Best Practices

- Use append() for adding single elements.
- Use extend() for combining lists.
- Use pop() when removed values are required.
- Use remove() when deleting by value.
- Avoid modifying a list while iterating over it.
- Use copy() when independent list modification is required.
- Prefer built-in sorting methods instead of implementing sorting algorithms manually.
