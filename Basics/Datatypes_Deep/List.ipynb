# List (Python)

A **list** is a built-in sequence data type in Python used to store an ordered collection of items.

## Characteristics

- Sequence data type.
- Stores a collection of multiple values in a single variable.
- Insertion order is preserved.
- Duplicate elements are allowed.
- Heterogeneous data types are allowed (e.g., `int`, `str`, `float`, `list`, `dict`, etc.).
- Dynamic in size (can grow or shrink at runtime).
- Elements are enclosed in **square brackets `[]`** and separated by commas.
- Mutable (elements can be added, removed, or modified after creation).
- Supports positive and negative indexing.
- Supports slicing to access a range of elements.
- Can be nested (a list can contain other lists).
- Allows repeated values and `None` as elements.

## Creating a List
This section shows common list creation patterns in Python: literals, mixed data types, empty lists, range conversion, and string splitting.

```python
# define some sample lists
numbers = [10, 20, 30, 40]
names = ["Alice", "Bob", "Charlie"]
mixed = [10, "Python", 3.14, True]
empty = []

print("Numbers:", numbers)
print("Names:", names)
print("Mixed:", mixed)
print("Empty:", empty)
```

```python
# input() reads text from the user.
# We expect the user to enter a Python list like [1, 2, 3].
# eval() converts that text into a real Python list object.
# Warning: eval can execute arbitrary code, so only use it with trusted input.
elements = eval(input("Enter Elements as a Python list, e.g. [1,2,3]: "))
print("User entered list:", elements)
```

```python
# Create a list from a range of numbers: 0 through 9.
lis_ = list(range(0, 10))
print("Range list:", lis_)
```

```python
# Convert a string into a list of words using split().
name = "Rohit Gore"
li = name.split()
print("Words list:", li)
```

## Accessing Elements of List
This section demonstrates how to get individual list items using positive and negative indexes.
    > By Using Index
        list follows zero based index. i.e. first element is zero.
        list supports both +ve and -ve indexes.
        +ve meant for left to right.
        -ve meant for right to left.

```python
demo_list = [10, 20, 30, 40, 50]
# Access elements by positive index (0-based) and negative index (from the end).
print("First element:", demo_list[0])
print("Fourth element:", demo_list[3])
print("Last element using negative index:", demo_list[-1])
```

    > By Using Slice Operators
        **syntax** -> list[start:stop:step]
        start -> where slice has to start, default 0
        stop -> where slice has to end, defalut len(list)
        step -> increment value, defalut 1


```python
demo_list = [1, 2, 3, 4, 5, 7, 89, 9]
print("Original list:", demo_list)
# Slice from index 0 to 4, stepping by 2.
print("Every second element in first 5 items:", demo_list[0:5:2])
# Slice from index 4 to the end.
print("Elements from index 4 to end:", demo_list[4:])
# Reverse slice from index 8 down to 5.
print("Reverse slice from position 8 to 5:", demo_list[8:4:-1])
```

## Traversing Elements of List
This section shows how to loop through list items using both while and for loops.

```python
li = [1, 2, 3, 4, 5, 675, 87, 2]

# Traverse using a while loop and explicit index.
i = 0
while i < len(li):
    print("Index", i, "value", li[i])
    i += 1

# Traverse using a for loop directly over elements.
for ele in li:
    print("Value:", ele)
```

## List Functions
This section explores built-in list functions and methods for inspecting and modifying lists.

```python
list_ex = [1, 2, 432, 54, 65, 87, 76, 1]
# dir() shows available methods for list objects.
print("List methods and attributes:", dir(list_ex))
```

```python
print("Length of list_ex:", len(list_ex))  # get the number of items in the list

print("Count of 1 in list_ex:", list_ex.count(1))  # count occurrences of a value
print("Count of 110 in list_ex:", list_ex.count(110))  # returns 0 if the value is not found

print("Index of first 1 in list_ex:", list_ex.index(1))  # returns first occurrence index
try:
    print("Index of 110 in list_ex:", list_ex.index(110))
except ValueError as e:
    print("ValueError:", e)  # 110 is not in the list
```

```python
list_ex = [1, 2, 432, 54, 65, 87, 76, 1]
list_ex.append(10000)  # add item to the end of the list
print("After append:", list_ex)

list_ex.insert(3, 3000)  # insert item at index 3 (0-based position)
print("After insert:", list_ex)

list_ex.extend([21])  # extend the list by appending all elements from another iterable
print("After extend:", list_ex)

list_ex.remove(1)  # remove the first occurrence of value 1
print("After remove(1):", list_ex)

last = list_ex.pop()  # remove and return the last item in the list
print("Popped value:", last)
print("List after pop:", list_ex)
```

#### Ordering Elements of List
This section covers list ordering methods: reversing the current list and sorting items.

```python
# Reverse the order of elements in-place.
list_ex = [1, 2, 432, 54, 65, 87, 76, 1]
list_ex.reverse()
print("Reversed list:", list_ex)
```

```python
# Sort the list in ascending order.
list_ex = [1, 2, 432, 54, 65, 87, 76, 1]
list_ex.sort()
print("Sorted ascending:", list_ex)

# Sort in descending order with reverse=True.
list_ex.sort(reverse=True)
print("Sorted descending:", list_ex)

# Mixed-type lists cannot be sorted, because Python cannot compare strings and integers.
list_ex = [1, 2, 432, 54, 65, 87, 76, "A"]
# The following line would raise a TypeError if executed:
# list_ex.sort()
print("Mixed list example (unsortable):", list_ex)
```

#### Aliasing and Cloning of List
This section shows the difference between two variables referencing the same list and how to create a true copy.

```python
# Aliasing: both variables refer to the same list object.
list_ex = [1, 2, 432, 54, 65, 87, 76, 1]
list_ex_2 = list_ex

print("Original list_ex:", list_ex)
print("list_ex_2 alias:", list_ex_2)

# Change via alias also changes the original list.
list_ex_2[1] = 1000

print("After modification, list_ex:", list_ex)
print("After modification, list_ex_2:", list_ex_2)
```

```python
# To avoid aliasing, create a copy of the original list.

# 1. Using slice notation to clone the list.
list_ex = [1, 2, 432, 54, 65, 87, 76, 1]
list_ex_2 = list_ex[:]

print("Original list_ex:", list_ex)
print("Cloned list_ex_2 using slice:", list_ex_2)

list_ex_2[1] = 1000
print("After modifying clone, original list_ex:", list_ex)
print("After modifying clone, list_ex_2:", list_ex_2)

# 2. Using the copy() method to clone the list.
print()
list_ex = [1, 2, 432, 54, 65, 87, 76, 1]
list_ex_2 = list_ex.copy()

print("Original list_ex:", list_ex)
print("Cloned list_ex_2 using copy():", list_ex_2)

list_ex_2[1] = 1000
print("After modifying clone, original list_ex:", list_ex)
print("After modifying clone, list_ex_2:", list_ex_2)
```

## List Comprehension
This section introduces list comprehension, a concise way to create new lists from existing iterables.
    syntax      list= [expression for item in list if condition]

```python
# ex 1
square_list = [x*x for x in range(1,10)]
print(square_list)

odd_list = [x for x in range(1,10) if x%2!=0]
print(odd_list)
```
