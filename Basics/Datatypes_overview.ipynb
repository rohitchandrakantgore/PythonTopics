# Datatypes in Python
## Fundatamental datatypes
- *Int*
- *Float*
- *Complex*
- *Bool*
- *Str*

## Derived Datatypes
- Bytes
- Bytearray
- Range
- List
- Tuple
- Set
- FrozenSet
- Dict
- None

```python
# Int Datatype
a=10
print(type(a))

# We can represent integers in different bases. For example, we can represent an integer in binary, octal, or hexadecimal format.
b=0b1010  # Binary
c=0o12    # Octal
d=0x1A    # Hexadecimal
print(a,b,c,d)
```

```python
# Float Datatype
#used to represent decimal numbers. It can also represent numbers in scientific notation.
# float values only used in decimal for only.
e=10.5
f=1.2e-3  # Scientific notation
print(type(e),type(f))
print(e,f)
```

```python
# Complex Datatype
#used to represent complex numbers. A complex number has a real part and an imaginary part.
g=2+3j
print(type(g))
print(g.real,g.imag)
```

```python
# Boolean Datatype
#used to represent truth values. It can have two possible values: True or False.
h=True
i=False
print(type(h),type(i))
print(h,i)
```

```python
# String Datatype
#used to represent text. A string is a sequence of characters enclosed in single quotes, double quotes, or triple quotes.
j='Hello'
k="World"
l='''This is a multi-line string.'''
print(type(j),type(k),type(l))
print(j,k,l)
```

```python
# Bytes Datatype
#used to represent a sequence of bytes. It is an immutable sequence of integers in the range 0 <= x < 256. 
# Bytes are often used to represent binary data, such as images

m=bytes([65, 66, 67])
print(type(m))
print(m)
```

```python
# ByteArray Datatype
#used to represent a mutable sequence of bytes. It is similar to the bytes datatype, but it can be modified after creation. 
# Bytearrays are often used when we need to manipulate binary
n=bytearray([65, 66, 67])
print(type(n))
print(n)
```

```python
# Range Datatype
#used to represent a sequence of numbers. It is often used in for loops to iterate over a sequence of numbers. 
# The range function generates a sequence of numbers from a starting value to an ending value, with an optional step value. The range function returns a range object, which is an immutable sequence of numbers.
o=range(5)
print(type(o))
print(list(o))  # Convert range object to list to see the numbers
```

```python
# List Datatype
#used to represent an ordered collection of items. A list can contain items of different datatypes
p=[1, 2, 3, 4, 5]
print(type(p))
print(p)
```

```python
# Tuple Datatype
#used to represent an ordered collection of items. 
# A tuple is similar to a list, but it is immutable, meaning that its contents cannot be changed after creation. 
# Tuples are often used to group related data together, such as coordinates or RGB values.
q=(1, 2, 3, 4, 5)
print(type(q))
print(q)
```

```python
# Set Datatype
#used to represent an unordered collection of unique items.
# Sets are often used to perform mathematical set operations, such as union, intersection, and difference
r={1, 2, 3, 4, 5}
print(type(r))
print(r)
```

```python
# FrozenSet Datatype
#used to represent an unordered collection of unique items, similar to a set.
# However, a frozenset is immutable, meaning that its contents cannot be changed after creation
s=frozenset([1, 2, 3, 4, 5])
print(type(s))
print(s)
```

```python
# Dictionary Datatype
#used to represent a collection of key-value pairs. A dictionary is an unordered collection of items
t={'a': 1, 'b': 2, 'c': 3}
print(type(t))
print(t)
```

```python
# None Datatype
#used to represent the absence of a value or a null value. 
# It is often used to indicate that a variable has not been assigned a value or that a function does not return a value.
u=None
print(type(u))
print(u)
```
