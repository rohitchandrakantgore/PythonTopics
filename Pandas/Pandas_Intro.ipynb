## Pandas Data Structures
- Series - 1D labeled array capable of holding data of any type (integer, string, float, python objects, etc.)
            can be created from a list, dict, or scalar value or data can loaded from a file (CSV, Excel, etc.)
            labels are called index and can be customized
- DataFrame - 2D labeled data structure with columns of potentially different types
            created by loading data from a file (CSV, Excel, etc.) or from a dict of Series objects or sql database
            can be created from lists, dicts, or numpy arrays or list of dicts.

```python
import pandas as pd
## Pandas Series
series = pd.Series([1, 2, 3, 4, 5])
print(series)
```

```python
import pandas as pd
## Pandas DataFrame
data = {'Name': ['Alice', 'Bob', 'Charlie'],
        'Age': [25, 30, 35],
        'City': ['New York', 'Los Angeles', 'Chicago']}
df = pd.DataFrame(data)
print(df)
```
