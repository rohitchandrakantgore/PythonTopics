## Pandas data structures
- Series: a 1D labeled array that can hold any data type. It is like a single column in a table.
- DataFrame: a 2D labeled table with rows and columns. It is the most commonly used pandas object.

### Key difference
- A Series is one-dimensional.
- A DataFrame is two-dimensional and contains rows and columns.

### Common ways to create them
- Series can be created from a list, dictionary, or scalar value.
- DataFrame can be created from a dictionary of lists, lists of dictionaries, CSV/Excel files, or other tabular data sources.

### Important idea
- Labels are called an index.
- Each column in a DataFrame can have a different data type.


````python
import pandas as pd

# Create a simple Series: a one-dimensional labeled array
series = pd.Series([1, 2, 3, 4, 5])
print(series)
print("\nType:", type(series).__name__)  # Show the object type
print("\nIndex:", series.index)         # Show the labels for values
print("Values:", series.values)         # Show the underlying values
````

````python
import pandas as pd

# Create a DataFrame from a dictionary of lists
# A DataFrame is a two-dimensional table with rows and columns
data = {
    'Name': ['Alice', 'Bob', 'Charlie'],
    'Age': [25, 30, 35],
    'City': ['New York', 'Los Angeles', 'Chicago']
}
df = pd.DataFrame(data)
print(df)
print("\nShape:", df.shape)                # number of rows and columns
print("\nColumns:", df.columns.tolist())    # list of column names
print("\nFirst row:")
print(df.iloc[0])                           # access the first row by position
````
