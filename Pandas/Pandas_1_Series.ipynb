# Pandas Series
A Series is a one-dimensional labeled array in pandas. It is useful for storing a single column of data or a sequence of values.

For the full pandas study roadmap, see the shared guide in the pandas folder.


```python
import pandas as pd

# A Series is a one-dimensional labeled array
numbers = pd.Series([10, 20, 30, 40])
print(numbers)
print("\nIndex:", numbers.index)      # labels for the values
print("Values:", numbers.values)      # raw values stored in the Series
print("Sum:", numbers.sum())          # aggregate operation
print("Mean:", numbers.mean())        # another common aggregate
```

## Important Series concepts
- A Series has values and an index.
- You can create a Series from a list, dictionary, or scalar.
- Use Series for single-column data or simple one-dimensional operations.
- A Series can be accessed by position with `.iloc[]` or by label with `.loc[]`.
- For a full guide to the pandas topics to learn next, see the shared guide in the pandas folder.


## Creation
Series can be created from a list, dictionary, NumPy array, range, or a list comprehension. This is one of the most important beginner skills in pandas.


```python
import numpy as np
import pandas as pd

## Empty Pandas Series
empty_series = pd.Series()
print(empty_series)
print("\nEmpty series length:", len(empty_series))

## Series with custom index
custom_index_series = pd.Series([100, 200, 300], index=['a', 'b', 'c'])
print("\nCustom index series:")
print(custom_index_series)

## Series from numpy array
data_series = pd.Series(np.array([5, 10, 15]))
print("\nSeries from numpy array:")
print(data_series)

## Series from dictionary
data_dict = {'a': 1, 'b': 2, 'c': 3}
dict_series = pd.Series(data_dict)
print("\nSeries from dictionary:")
print(dict_series)

## Series from range
range_series = pd.Series(range(3))
print("\nSeries from range:")
print(range_series)

## Series with string data
string_series = pd.Series(['apple', 'banana', 'cherry'])
print("\nSeries with string data:")
print(string_series)

## Series with list comprehension
list_comp_series = pd.Series([x**2 for x in range(3)])
print("\nSeries from list comprehension:")
print(list_comp_series)
```

## Accessing elements of Series
- Labels do not need to be unique, but they must be hashable.
- You can access values by position using `.iloc[]`.
- You can access values by label using `.loc[]` or by directly using the label.


## 1. Accessing elements from a Series with position
- Use integer-based positions.
- Slice operations also work.


```python
### Series with mixed data types
mixed_series = pd.Series([1, 'two', 3.0, True])
print(mixed_series)

### Accessing first and last elements of a Series
first_element = mixed_series.iloc[0]
last_element = mixed_series.iloc[-1]
print("\nFirst element:", first_element)
print("Last element:", last_element)

# Accessing first 3 elements of the series
first_three_elements = mixed_series.iloc[:3]
print("\nFirst three elements:\n", first_three_elements)

## Accessing elements by step
step_elements = mixed_series.iloc[::2]
print("\nElements with step 2:\n", step_elements)
```

### 2. Accessing elements using labels


```python
# Series with custom index
custom_index_series = pd.Series([100, 200, 300], index=['a', 'b', 'c'])

# Accessing a single element
print(custom_index_series['a'])

# Accessing multiple elements using a list of indices
print(custom_index_series[['b', 'c']])
```

```python
# Inspecting a Series
# These methods help you understand the data quickly.
print(numbers.head())
print("\nTail:")
print(numbers.tail())
print("\nInfo:")
print(numbers.info)
print("\nDescribe:")
print(numbers.describe())

# Indexing and selection basics for a Series
print("\nValue at position 1:", numbers.iloc[1])
print("\nValue by label (if index is default integers):", numbers.loc[1])
print("\nSlice from index 1 to 3:")
print(numbers.iloc[1:4])
```
