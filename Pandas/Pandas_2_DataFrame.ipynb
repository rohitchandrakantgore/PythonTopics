## Pandas DataFrame
A DataFrame is a two-dimensional labeled table in pandas. It is used to store and manipulate tabular data in rows and columns.


```python
import pandas as pd

print("Pandas DataFrame basics")
```

# Creation of Data Frame

```python
## Empty DataFrame
empty_df = pd.DataFrame()
print(empty_df)
print("\nShape:", empty_df.shape)
```

```python
## DataFrame from a list
# A list of values creates a single-column DataFrame by default
list_data = [1, 'Alice', 25, 2, 'Bob', 30, 3, 'Charlie', 35]
print(pd.DataFrame(list_data))
```

```python
## DataFrame from a dictionary
# This is one of the most common ways to create a DataFrame
# Each key becomes a column name, and the values become the column data
data = {'Name': ['Alice', 'Bob', 'Charlie'],
        'Age': [25, 30, 35]}
df = pd.DataFrame(data)
print(df)
print("\nColumns:", df.columns.tolist())
print("\nRows:", len(df))
```

```python
## DataFrame from a NumPy array
import numpy as np

# A 2D array can be converted into a DataFrame with named columns
# This is useful when working with numeric data from NumPy
data_array = np.array([[1, 2, 3], [4, 5, 6]])
df_from_array = pd.DataFrame(data_array, columns=['A', 'B', 'C'])
print(df_from_array)
```

```python
## DataFrame from a CSV file
from pathlib import Path

# read_csv() is one of the most important pandas functions
# It reads a CSV file into a DataFrame for analysis
csv_candidates = [Path('sample_employees.csv'), Path('Pandas/sample_employees.csv')]
csv_path = next((p for p in csv_candidates if p.exists()), Path('sample_employees.csv'))

data = pd.read_csv(csv_path)
print(data)
print("\nShape:", data.shape)
```

## DataFrame index
The index is the row label of a DataFrame. It helps identify rows and can be customized.


```python
# Show first 5 rows (preview)
print("Head:")
print(df.head())

# Show last 5 rows
print("\nTail:")
print(df.tail())

# Show concise summary of DataFrame (dtypes, non-null counts)
print("\nInfo:")
df.info()

# Show basic statistics for numeric columns
print("\nDescribe:")
print(df.describe())

# Show shape (rows, columns)
print("\nShape:", df.shape)
```

```python
# Show column names as list
print("\nColumns:", df.columns.tolist())

# Select one column (Series)
print("\nSelect one column 'Name':")
print(df['Name'])

# Select rows by integer position (iloc)
print("\nSelect rows 0 to 1 using iloc:")
print(df.iloc[0:2])

# Select rows by label (loc) - here index labels 0 to 1
print("\nSelect rows 0 to 1 using loc:")
print(df.loc[0:1])

# Boolean filtering: rows where Age > 25
print("\nBoolean filter: Age > 25")
print(df[df['Age'] > 25])
```
