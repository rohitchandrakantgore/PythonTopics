# First Class Functions
in python functions are treated as first class objects.
they can be used just like numbers.

we can
- Assign Functions to variables
- Pass them as arguments to other funcstions
- return them from functions
- store them into data structures like list or dictionies


```python
# Assigning functions to variables

def hello(name):
    return f"Hello, {name}"

#assigning function to a variable
nm = hello
#calling the function using the variable
print(nm("Rohit"))
```

```python
# Passing functions as arguments to other functions
def greet(name):
    return f"Hello, {name}!"

def farewell(name):
    return f"Goodbye, {name}!"

def call_function(func, arg):
    return func(arg)

print(call_function(greet, "Alice"))
print(call_function(farewell, "Bob"))
```

```python
# Returning functions from other functions
def outer_function():
    def inner_function():
        return "Hello from the inner function!"
    return inner_function

my_inner_function = outer_function()
print(my_inner_function())
```

```python
# Storing functions in data structures

def add(x, y):
    return x + y

def subtract(x, y):
    return x - y

# Storing functions in a list
operations = [add, subtract]

# Using the functions from the list
result1 = operations[0](5, 3)  # Calls add(5, 3)
result2 = operations[1](5, 3)  # Calls subtract(5, 3)

print(result1)  # Output: 8
print(result2)  # Output: 2

# storing functions in a dictionary
def multiply(x, y):
    return x * y

def divide(x, y):
    if y != 0:
        return x / y
    else:
        return "Cannot divide by zero"  

operations_dict = {
    "add": add,
    "subtract": subtract,
    "multiply": multiply,
    "divide": divide
}   

# Using the functions from the dictionary
result3 = operations_dict["multiply"](4, 5)  # Calls multiply(4, 5)
result4 = operations_dict["divide"](10, 2)   # Calls divide(10, 2)      

print(result3)  # Output: 20
print(result4)  # Output: 5.0
```
