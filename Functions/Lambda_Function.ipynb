# Lambda Functions
small anonymous functions means they do not have a defined name.

these are small, short-lived functions used to pass simple logic to another functions.

- contains only one experssion
- return of that expression is returned automatically (no return keyword needed.)

Syntax:
lambda argument:expression

```python
name = "Rohit"
upper = lambda name: name.upper()
print(upper(name))
```

```python
# condition checking using lambda function
check = lambda x: "Genz" if x < 25 else "Millenial"
print(check(20))  # Output: Genz
print(check(30))  # Output: Millenial
```

```python
# list comprehension with lambda function
numbers = [lambda arg = x: arg * 2 for x in range(5)]
for i in numbers:
    print(i())  # Output: 0, 2, 4, 6, 8
```

```python
# returning multiple results from a lambda function
# although lambda functions are typically used for single expressions, you can return multiple values as a tuple. Here's an example:
calc = lambda x, y: (x + y, x - y, x * y, x / y)
result = calc(10, 5)
print(result)  # Output: (15, 5, 50, 2.0)
```

```python
#Length of each string in list
stocks = ['playway', 'boombit', 'cd projekt']
length = list(map(lambda stock: len(stock), stocks))
print(length)
```

```python
def sort_list(items):
    return sorted(items, key=lambda item: item[0])

print(sort_list(['playway', 'boombit', 'cd projekt']))
```

```python
items = [(3, 4), (2, 5), (1, 4), (6, 1)]
items.sort(key=lambda item: item[0]**2 + item[1]**2)
print(items)
```

```python
stocks = [
    {'index': 'mWIG40', 'name': 'TEN', 'price': 304},
    {'index': 'mWIG40', 'name': 'PLW', 'price': 309},
    {'index': 'sWIG80', 'name': 'BBT', 'price': 22}
]

stocks.sort(key=lambda item: item['price'])
print(stocks)
```

```python
stocks = [
    {'index': 'mWIG40', 'name': 'TEN', 'price': 304},
    {'index': 'mWIG40', 'name': 'PLW', 'price': 309},
    {'index': 'sWIG80', 'name': 'BBT', 'price': 22}
]

print(list(filter(lambda item: item['index'] == 'mWIG40', stocks)))
```

```python
items = ['P-1', 'R-2', 'D-4', 'F-6']
print(list(map(lambda item: item.replace('-', ''), items)))
```

```python
num1 = [4, 2, 6, 2, 11]
num2 = [5, 2, 3, 3, 9]

print(list(map(lambda x, y: x % y, num1, num2)))
```
